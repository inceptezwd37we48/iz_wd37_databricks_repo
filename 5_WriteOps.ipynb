
## mode - append , overwrite , errorifexists, ignore 
default mode -> errorifexists

overwrite - it will remove the exisitng data overwrite with new data 

append - adding new data , adding  a new file 

ignore - do nothing 



```python
ddl_str="custid int, fname string, lname string, age int, profession string"
cust_df=spark.read.schema(ddl_str).csv("/Volumes/izwd37dev/wd37db/rawdatta/custs")
cust_df.write.mode("ignore").csv("/Volumes/izwd37dev/wd37db/rawdatta/csvout",sep="|",header=True)
```

```python
cust_df.write.mode("append").csv("/Volumes/izwd37dev/wd37db/rawdatta/csvout",sep="|",header=True)
```

# Load Data into Table  -> Lakhouse formation 
## Data lake + warehouse -> Lakehouse 
## cloud storage (s3)   + databricks sql  ( delta )  -> lakehouse 

```python
ddl_str="custid int, fname string, lname string, age int, profession string"
cust_df=spark.read.schema(ddl_str).csv("/Volumes/izwd37dev/wd37db/rawdatta/custs")

# writing into the table 
cust_df.write.saveAsTable("izwd37dev.wd37db.tbl_customer")
```

```python
# read the table 

# using DSL approach 

cust_tbl_df=spark.read.table("izwd37dev.wd37db.tbl_customer")

cust_tbl_df.show(5)
cust_tbl_df.printSchema()

cust_tbl_df.count()


```

```python
cust_tbl_df=spark.sql("select * from izwd37dev.wd37db.tbl_customer")
cust_tbl_df.show(5)
cust_tbl_df.printSchema()

cust_tbl_df.count()
```
