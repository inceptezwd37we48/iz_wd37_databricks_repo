```python
from datetime import datetime
def log_detail(fn):
    def wrapper():
        print(f"INFO : {datetime.now()} function started running ")
        fn()
        print(f"INFO : {datetime.now()} function completed running ")
    return wrapper

    
```

```python
@log_detail
def print_detail():
    print("welcome all to declartive world")
    print("This is day 1 class ")
    print("Today topic more on theory plus essentials understanding ")
```

```python
print_detail()
```

```python
@log_detail
def all_declrative_name():
    print("1.DLT (delta live Table)")
    print("2. declarative pipelines")
    print("3. SPARK Declarative pipelines(SDP)")



all_declrative_name()
```
